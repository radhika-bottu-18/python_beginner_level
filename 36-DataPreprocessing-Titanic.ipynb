```python
import matplotlib.pyplot as plt 
from matplotlib import style 

males = (titanic['Sex'] == 1).sum()
print(males)
females = (titanic['Sex'] == 2).sum()
print(females)
p = [males,females ]
plt.pie(p ,labels = ['Male','Female'], 
       colors= ['green','yellow'],explode=(0.10,0),
       startangle=0)
plt.axis('equal')
plt.show()
```
