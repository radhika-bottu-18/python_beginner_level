```python
def count(lst):
    even =0
    odd=0
    for i in lst:
        if i %2== 0:
            even+=1
        else:
            odd+=1
    return even,odd

lst = [3,4,5,6,8,10,12,9,7]
even,odd = count(lst)
print('even=',even)
print('odd=',odd)

```

```python
def fib(n):
    a=0
    b=1
    print(a)
    print(b)
    for i in range(2,n):
        c=a+b
        a=b
        b=c
        print(c)
fib(7)
```

```python
def factorial(n):
    ans=1
    for i in range(1,n+1):
        ans *= i
    print(ans)

factorial(3)
```

```python
def fact(n):
    if(n ==1):
        return 1
    else:
        return n*fact(n-1)
fact(5)
```
