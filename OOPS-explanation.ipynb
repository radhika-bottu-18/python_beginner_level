```python
class computer : 
    def __init__(self):
        self.name = 'prakash'
        self.age = 39
    def update(self):
        self.age = 40 

com1 = computer()
com2 = computer()

print(com1.name,com1.age)
print(com2.name,com2.age)
com2.update()

print(com1.name,com1.age)
print(com2.name,com2.age)
```
