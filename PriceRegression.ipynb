```python
def get_average(df,column):
    return sum(df[column])/len(df[column])
```

```python
def get_average_between_two_columns(df,column1,column2):
    uniqueRegionList = list(df[column1].unique())
    averageperregion = []
    for region in uniqueRegionList :
        regiondata=df[df[column1] == region]
        region_average= get_average(regiondata,column2)
        averageperregion.append(region_average)

    df_region_avgprice = pd.DataFrame({'region': uniqueRegionList,'column2': averageperregion})
    col1_col2_sorted_index = df_region_avgprice.column2.sort_values(ascending=True).index.values
    col1_col2_sorted_data = df_region_avgprice.reindex(col1_col2_sorted_index)

    return col1_col2_sorted_data
```

```python
def plot(data,xlabel,ylabel):
    plt.figure(figsize=(15,15))
    ax=sns.barplot(x=data.region,y=data.column2,palette='rocket')
    plt.xticks(rotation=90)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.title(('Average '+ylabel+' of Avacoda According to '+xlabel))
```

```python
data1 = get_average_between_two_columns(df,'region','AveragePrice')
plot(data1,'Region','Price ($)')
```

```python
from sklearn.model_selection import train_test_split
x= df.drop(['AveragePrice'],axis=1)
y=df['AveragePrice']

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state=0)
print('Training data :',x_train.shape,'-', y_train.shape[0], ' samples')
print('Testing data :',x_test.shape,'-', y_test.shape[0], ' samples')
```

```python
from sklearn.linear_model import LinearRegression 
# bulid and fit the model
model = LinearRegression()
model.fit(x_train,y_train)
```

```python
# prediction and calculate the accuracy for the testing dataset
test_pre = model.predict(x_test)
test_score= r2_score(y_test,test_pre)
print('the accuracy of the model is = ',test_score)
```
